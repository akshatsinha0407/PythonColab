```python
class dummyPromptTemplate:
  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def format(self, input_dict):
    return self.template.format(**input_dict)
```

```python
template = dummyPromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=['length','topic']
)
```

```python
template.format({'length':'short','topic':'india'})
```

```python
class dummyLLMChain:
  def __init__(self, llm, prompt):
    self.llm = llm
    self.prompt = prompt

  def run(self, input_dict):
    final_prompt = self.prompt.format(input_dict)
    result = self.llm.predict(final_prompt)
    return result['response']
```

```python
template = dummyPromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=['length','topic']
)
```

```python
llm = dummyLLM()
```

```python
chain = dummyLLMChain(llm,template)
```

```python
chain.run({'length':'short','topic':'india'})
```

Makada kisi me format kisi me predict , aager ek chian se do kam karana also not possible , So runnables comes in picture

```python
from abc import ABC , abstractmethod
```

```python
class Runnable(ABC):
  @abstractmethod
  def invoke(input_data):
    pass
```

```python
import random


class dummyLLM2:
  def __init__(self):
    print("LLM Created")

  def invoke(self, prompt):
    response_list1 = [
      "Aksmmkmmkmkm",
      "huihuihui",
      "chacha"
    ]

    return {'response': random.choice(response_list1)}

  def predict(self, prompt):
    response_list = [
      "Aksmmkmmkmkm",
      "huihuihui",
      "chacha"
    ]

    return {'response': random.choice(response_list)}
```

```python
class dummyPromptTemplate2(Runnable):
  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def invoke(self, input_dict):
    return self.template.format(**input_dict)

  def format(self, input_dict):
    return self.template.format(**input_dict)
```

```python
class RunnableConnector(Runnable):

    def __init__(self, runnable_list):
        self.runnable_list = runnable_list

    def invoke(self , input_data):
        for runnable in self.runnable_list:
            runnable.invoke(input_data)
        return input_data


```

```python
template = dummyPromptTemplate(
    template="Write a {length} poem about {topic}",
    input_variables=['length','topic']
)
```

```python
llm = dummyLLM2()
```

```python
chain2=RunnableConnector([llm,template])
```

```python
chain2.invoke({'length':'short','topic':'india'})
```
